# Transparent synthetic stress simulation
All profiles/labels are invented. Results measure learning an invented mechanism, not real-person accuracy. Production continues to use the protected real-data models.

```python
from pathlib import Path
import json, pandas as pd
E=Path.cwd()
if not (E/'protocol.json').exists():
    E=E/'experiments'/'synthetic_simulation_20261026'
load=lambda name: json.loads((E/name).read_text())
protocol=load('protocol.json')
print(protocol['purpose'])
```

## 1. Inputs and label generation
The19core inputs use the project schema. Shared latent factors create correlated hypothetical profiles. The generator fixes both its label equation and uncertainty before fitting any model. The latent score is not a training input.

```python
print(protocol['features'])
print(protocol['label_equation'])
df=pd.read_csv(E/load('data_manifest.json')['csv'])
print(df.groupby(['split','target']).size().unstack())
assert df.sample_id.is_unique
assert not df.duplicated(protocol['features']).any()
```

## 2. Preprocessing and split discipline
Numeric columns use StandardScaler; categorical columns use a fixed one-hot schema. Each preprocessor is fit on the model-fitting rows only. Independent development rows select neural epochs; validation chooses decision rules; the synthetic test is evaluated after all choices are frozen.

```python
print(protocol['split'])
print(pd.read_csv(E/'training_roles.csv').role.value_counts())
print(protocol['fitting'])
```

## 3. All six models
The existing families and architectures are retained. Neural models use two predetermined seeds and average probabilities. They operate across fixed tabular feature positions, not a time series.

```python
v=load('validation_metrics.json')
for name,obj in v.items():
    print(name, 'seed accuracies:', [round(s['accuracy'],4) for s in obj['per_seed_raw']], 'policy feasible:', obj['policy_feasible'])
```

## 4. Frozen evaluation
Report raw argmax and the validation-selected operating rule separately. The test is synthetic; no BRFSS holdout was inspected for this experiment. A high score on an easy generator cannot establish real-world accuracy.

```python
results=pd.read_csv(E/'SYNTHETIC_comparison.csv')
print(results.to_string(index=False))
test=load('SYNTHETIC_test_metrics.json')
print('Synthetic majority baseline:', test['majority_baseline']['accuracy'])
print('Real-world accuracy:',test['real_world_accuracy'])
print(load('verification.json'))
```
